`Statevector(M)` defines vector that describes the state of a quantum system.
```{python}
from qiskit.quantum_info import Statevector
from numpy import sqrt
from qhelpers import qdisplay

u = Statevector([1 / sqrt(2), 1 / sqrt(2)])
display(u.draw("latex"))
```

`.is_valid()` checks whether vector is a valid quantum state vector (simm. if the euclidian norm == 1).
```{python}
w = Statevector([1 / 3, 2 / 3])
display(u.is_valid())
display(w.is_valid())
```

`.measure()` simulates a standard basis measurement.
```{python}
outcome, state = u.measure()
print(f"Measured: {outcome}\nPost-measurement state:")
display(state.draw("latex"))
```

`Statevector` comes with a `.sample_counts(COUNT)` that simulates any number of measurements on the system, each time starting with a fresh copy of the state.
```{python}
from qiskit.visualization import plot_histogram

statistics = u.sample_counts(1000)
plot_histogram(statistics)
```

`Operator` defines a unitary operator, also includes `.draw("TYPE")` (which is simm. to array_to_latex()) like `Statevector`.
```{python}
from qiskit.quantum_info import Operator

# operator is for unitary operations
T = Operator([[1, 0], [0, (1 + 1.0j) / sqrt(2)]])
H = Operator([[1 / sqrt(2), 1 / sqrt(2)], [1 / sqrt(2), -1 / sqrt(2)]])

qdisplay(T.draw("latex"))
```

`.evolve(OPERATOR)` is to apply a unitary operation to a `Statevector`.
```{python}
u.evolve(T)
u.evolve(H)

display(u.draw("latex"))
```

`mpl` stands for MatPlotLib.
```{python}

circuit = QuantumCircuit(1)

circuit.h(0)
circuit.t(0)

```
```{python}
qdisplay(Operator.from_circuit(circuit).draw("latex"))
```
